# Exploração da PAM — IBGE/SIDRA
Este notebook documenta a coleta inicial e as validações de cobertura, unidades e marcadores. A análise só deve prosseguir com uma extração oficial válida.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').exists())
data_path = ROOT / 'data' / 'processed' / 'pam_municipal_tratada.csv'
if not data_path.exists():
    raise FileNotFoundError('Execute primeiro python src/collect_pam.py; a API oficial ainda não foi coletada.')
df = pd.read_csv(data_path, dtype={'codigo_ibge': str})
df.head()
```

```python
display(df.groupby('municipio').agg(linhas=('ano','size'), anos=('ano','nunique'), produtos=('produto','nunique'), variaveis=('variavel','nunique')))
display(df.groupby(['municipio','ano']).size().unstack())
display(df.groupby(['variavel','unidade'], dropna=False).size().sort_values(ascending=False).head(20))
```
