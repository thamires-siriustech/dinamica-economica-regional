# Análise preliminar da base agrícola secundária
**Aviso:** estes números vêm de páginas secundárias que declaram IBGE/PAM como fonte. Eles ainda não foram reconciliados linha a linha com o SIDRA e não devem ser confundidos com extração oficial validada.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'data').exists())
df = pd.read_csv(ROOT / 'data/processed/pam_base_secundaria_piloto.csv', encoding='utf-8-sig')
assert len(df) == 66
assert not df.duplicated(['codigo_ibge','ano','cultura']).any()
df.head()
```

```python
resumo = df.query('ano == 2025').pivot(index='municipio', columns='cultura', values='producao_t')
display(resumo)
for cultura, grupo in df.groupby('cultura'):
    ax = grupo.pivot(index='ano', columns='municipio', values='producao_t').plot(marker='o', title=f'Produção: {cultura} — base secundária')
    ax.set_ylabel('Toneladas'); ax.grid(alpha=.25); plt.show()
```

## Interpretação
Não atribuir as mudanças de produção apenas à área ou produtividade sem decompor as variáveis. Valores em reais são nominais. Reconciliar área, produção, rendimento e valor de cada linha com SIDRA antes de usar os achados como resultados oficiais.
