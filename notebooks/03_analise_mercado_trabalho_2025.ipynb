# Mercado de trabalho formal — piloto 2025
Os números são transcritos de fonte secundária, ainda não reconciliados com microdados oficiais do MTE. O objetivo é demonstrar validações e análise descritiva.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'data').exists())
data = ROOT / 'data/processed'
municipios = pd.read_csv(data / 'caged_resumo_municipal_2025_secundario.csv', encoding='utf-8-sig')
setores = pd.read_csv(data / 'caged_setor_2025_secundario.csv', encoding='utf-8-sig')
auditoria = pd.read_csv(data / 'auditoria_fontes_saldo_lucas_2025.csv', encoding='utf-8-sig')
assert (municipios.admissoes - municipios.desligamentos == municipios.saldo).all()
display(municipios[['municipio','admissoes','desligamentos','saldo']])
```

```python
display(setores.pivot(index='setor_agregado', columns='municipio', values='saldo'))
display(auditoria[['fonte','saldo_empregos','status','nota_metodologica']])
ax = municipios.sort_values('saldo').plot.barh(x='municipio', y='saldo', legend=False, title='Saldo divulgado — fonte secundária')
ax.set_xlabel('Admissões menos desligamentos'); plt.show()
```

## Limites de interpretação
Saldo de movimentações não é estoque de empregos nem taxa de desemprego. A divergência de +1.803 versus +1.767 para Lucas do Rio Verde segue sem explicação confirmada. Conciliar microdados, competências e revisões do Novo CAGED antes de conclusão definitiva.
